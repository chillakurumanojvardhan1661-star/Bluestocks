# Day 1: Project Setup + Data Ingestion (ETL)
**Mutual Fund Analytics Platform - Bluestock Fintech**

This notebook demonstrates the ingestion and initial validation of the 10 mutual fund datasets provided for the capstone project, as well as the custom script that fetches live/historical NAV data from the public `mfapi.in` API.

```python
import pandas as pd
import numpy as np
import os

RAW_DIR = "../data/raw"
print("Available raw datasets:", os.listdir(RAW_DIR))
```

### Load All 10 CSV Datasets
We will load each CSV using Pandas and inspect its shape, columns, and a few sample rows.

```python
datasets = {}
for filename in sorted(os.listdir(RAW_DIR)):
    if filename.endswith('.csv') and not filename.startswith('live_nav'):
        name = filename.split('.')[0]
        path = os.path.join(RAW_DIR, filename)
        datasets[name] = pd.read_csv(path)
        print(f"{name}: Shape = {datasets[name].shape}, Columns = {datasets[name].columns.tolist()}")
```

### Validate AMFI Codes
Checking if all scheme codes in `01_fund_master` have corresponding NAV history in `02_nav_history`.

```python
fund_master = datasets['01_fund_master']
nav_history = datasets['02_nav_history']

master_codes = set(fund_master['amfi_code'].unique())
nav_codes = set(nav_history['amfi_code'].unique())

missing_in_nav = master_codes - nav_codes
print(f"AMFI codes in fund_master: {len(master_codes)}")
print(f"AMFI codes in nav_history: {len(nav_codes)}")
print(f"Missing codes in nav_history: {missing_in_nav}")
```

### Understanding Fund Master Metadata
Let's see some distributions of fund houses, categories, sub-categories, and risk grades.

```python
print("--- Unique Fund Houses ---")
print(fund_master['fund_house'].value_counts())
print("\n--- Unique Categories ---")
print(fund_master['category'].value_counts())
print("\n--- Unique Risk Categories ---")
print(fund_master['risk_category'].value_counts())
```
