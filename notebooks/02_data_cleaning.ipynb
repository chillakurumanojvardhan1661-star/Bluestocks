# Day 2: Data Cleaning + SQL Database Design
**Mutual Fund Analytics Platform - Bluestock Fintech**

This notebook demonstrates the cleaning of the datasets, filling in missing dates in the NAV history (using forward fill), defining the SQLite relational star schema, and loading the cleaned records into the database.

```python
import os
import sqlite3
import pandas as pd
import numpy as np

DB_PATH = "../data/db/bluestock_mf.db"
conn = sqlite3.connect(DB_PATH)
print("Connected to SQLite database successfully!")
```

### Verification of Loaded Tables
Let's see the list of tables in our SQLite database and print their record counts.

```python
tables = ['dim_fund', 'dim_date', 'fact_nav', 'fact_transactions', 'fact_performance', 'fact_aum', 'fact_portfolio', 'fact_sip_industry']
for t in tables:
    count = pd.read_sql_query(f"SELECT COUNT(*) FROM {t}", conn).iloc[0, 0]
    print(f"Table '{t}': {count} rows")
```

### Sample Queries Execution
Let's execute query 1 (Top 5 funds by AUM) and query 4 (Total transaction amount by state) to check the integrity of loaded tables.

```python
print("=== Top 5 Funds by AUM ===")
df_aum = pd.read_sql_query("""
    SELECT amfi_code, scheme_name, fund_house, category, aum_crore
    FROM fact_performance
    ORDER BY aum_crore DESC
    LIMIT 5;
""", conn)
display(df_aum)

print("\n=== Total Investment by State ===")
df_state = pd.read_sql_query("""
    SELECT state, COUNT(*) AS tx_count, SUM(amount_inr) AS total_invested_inr
    FROM fact_transactions
    GROUP BY state
    ORDER BY total_invested_inr DESC
    LIMIT 5;
""", conn)
display(df_state)
```
